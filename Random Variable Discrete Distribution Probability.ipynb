```python
unique=np.unique(df.values)
from collections import Counter
arr=df.values.reshape(36,)
# cnt=Counter(arr)

freq={}
for i in arr:
    if i not in freq:
        freq[i]=1
    if i in freq:
        freq[i]= freq[i]+1

sort=sorted(freq.items() ,key=lambda item:item[1] ,reverse=True)
print("Uniq values      :{}\nFreq unsorted    :{}\nFreq Sorted      :{}".format(unique,freq,sort))
```
